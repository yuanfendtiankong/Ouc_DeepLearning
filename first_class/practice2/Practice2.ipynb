### I. Background

A small network can be treated as a box of knobs: gradient descent turns them step by step until the curve the box produces fits the data. How well a curve fits is measured by the loss — the average of the squared errors, and smaller is better. In this practice the same 20 points are fitted in three ways: a straight line, a polynomial, and the box of knobs; then the box is given more knobs, and the three losses are compared.

## Question 1 · Make the points and fit two curves

Build the 20 noisy points, then fit a straight line and a degree-6 polynomial, and record the two losses.
`loss_of` is given: it returns the average of the squared errors.


```python
# 🟢 EDIT: set SID to the last two digits of your student ID
SID = 23

import numpy as np, torch, torch.nn as nn
torch.manual_seed(42)

x = np.linspace(-3, 3, 20)
y = 2 * x ** 2 - 3 * x + 1 + np.random.default_rng(1000 + SID).normal(0, 1.5, size=x.shape)


def loss_of(pred, tgt):
    """How far off a curve is: the average of the squared errors."""
    return float(((pred - tgt) ** 2).mean())


def fit_and_loss(x, y, k):
    """Fit a degree-k curve to the points, then return how far off it is."""
    return loss_of(np.polyval(np.polyfit(x, y, k), x), y)


# 🟢 EDIT: call fit_and_loss twice — once with degree 1, once with degree 6
# YOUR CODE HERE
loss_line = fit_and_loss(x,y,1)
loss_poly = fit_and_loss(x,y,6)

print("y ranges from %.2f to %.2f" % (y.min(), y.max()))
print("straight line loss =", round(loss_line, 3))
print("polynomial loss =", round(loss_poly, 3))
```

## Question 2 · The point of this practice: 1, 4 or 16 knobs?

Change one number, run it, and write the three losses down.
The code below is complete: treat the network as a box of knobs that gradient descent tunes step by step.


```python
# YOUR CODE HERE
raise NotImplementedError()
```

```python
# 🟢 EDIT: pick three numbers of knobs (1, 4, 16 is a good choice)
# YOUR CODE HERE
KNOBS = [1,4,16]

def fit_small_net(hidden, steps=200, seed=42):
    """hidden = number of knobs. Returns (loss, the curve it produces)."""
    torch.manual_seed(seed)
    net = nn.Sequential(nn.Linear(1, hidden), nn.ReLU(), nn.Linear(hidden, 1))
    opt = torch.optim.Adam(net.parameters(), lr=0.05)
    lossf = nn.MSELoss()
    xt = torch.tensor(x, dtype=torch.float32).view(-1, 1)
    yt = torch.tensor(y, dtype=torch.float32).view(-1, 1)
    for _ in range(steps + 1):
        opt.zero_grad(); loss = lossf(net(xt), yt); loss.backward(); opt.step()
    with torch.no_grad():
        xs = torch.tensor(np.linspace(-3, 3, 300), dtype=torch.float32).view(-1, 1)
        return float(lossf(net(xt), yt).detach()), net(xs).numpy().ravel()

results = []
for h in KNOBS:
    l, curve = fit_small_net(h)
    results.append((h, round(l, 3)))
    print("%2d knobs -> loss = %.3f" % (h, l))

print("\nyour three numbers:", results)
```

## Question 3 · Plot the three curves in one figure

Draw the 20 points and the three fitted curves together, and save the figure.


```python
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

plt.rcParams["font.sans-serif"] = ["DejaVu Sans"]
plt.rcParams["axes.unicode_minus"] = False

plt.figure(figsize=(6.4, 3.4))
plt.scatter(x, y, color="#12324F", s=26, label="20 points")
colors = ["#C0392B", "#B8860B", "#2E8B57", "#1F4E79"]
for (h, _), c in zip(results, colors):
    l, curve = fit_small_net(h)
    plt.plot(np.linspace(-3, 3, 300), curve, color=c, lw=2,
             label="%d knobs, loss %.3f" % (h, l))
plt.xlabel("x"); plt.ylabel("y")
plt.legend(fontsize=8)
plt.title("More knobs = more bends — SID %d" % SID)

# 🟢 EDIT: save the figure as fit_curves_<SID>.png
# YOUR CODE HERE

plt.tight_layout()
plt.savefig(f"fit_curves_{SID}.png")
plt.show()
```
