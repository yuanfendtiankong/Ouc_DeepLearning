## Question 2 · Draw a line chart and save it

Plot the noisy curve and save the figure; give the title your own words (keep `SID` in it).


```python
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

plt.rcParams["font.sans-serif"] = ["DejaVu Sans"]
plt.rcParams["axes.unicode_minus"] = False

x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x) + rng.normal(0, 0.15, size=x.shape)   # the noise follows your SID

plt.figure(figsize=(6, 3.2))
plt.plot(x, y, color="#1F4E79", lw=1.8)
plt.title("My First Plot — SID %d" % SID)
plt.xlabel("x")
plt.ylabel("sin(x) + noise")

# 🟢 EDIT: name the file warmup_line_<SID>.png and save the figure
# YOUR CODE HERE
fname = f"warmup_line_{SID}.png"
plt.savefig(fname)

plt.tight_layout()
print("saved:", fname)
plt.show()
```
