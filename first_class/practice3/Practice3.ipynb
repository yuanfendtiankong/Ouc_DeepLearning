### I. Background

A model never sees "the dataset"; it sees one batch at a time. Three words describe the path from files to a model: **Dataset** (how large it is, and what the i-th sample looks like), **DataLoader** (splits it into batches and shuffles them), **batch** (how much is fed at a time). In this practice you load MNIST — 60,000 grayscale images of 28 x 28 handwritten digits — and follow one image and one batch along that path.

## Question 1 · Load MNIST and look at one image

MNIST is already on the platform, so `download=False`. Count the samples, then read one image and its label.


```python
# 🟢 EDIT: set SID to the last two digits of your student ID
SID = 23

import math, numpy as np
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

# 🟢 EDIT: Optional parameter:download=False
ds = datasets.MNIST(root="./data", train=True, download=False, transform=transforms.ToTensor())

# 🟢 EDIT: count the samples and read the first image with its label. You may use len(ds) for the number of samples, and img, label = ds[0] for the first one
# YOUR CODE HERE
n_total = len(ds)
img, label = ds[0]

print("len(ds) =", n_total)
print("shape of ds[0] =", img.shape, " label =", label)
```

## Question 2 · Show one image large and print all its pixels

Pick an image with your own index, show it large and save it, then print all of its pixel values (28 × 28 numbers, 0-255).


```python
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
plt.rcParams["font.sans-serif"] = ["DejaVu Sans"]
plt.rcParams["axes.unicode_minus"] = False

# 🟢 EDIT: pick one image to look at (an index from 0 to 59999). You may use SID % len(ds) to make the index follow your SID
# YOUR CODE HERE
idx = SID % n_total

im, lb = ds[idx]
print("image %d: shape %s, label %d" % (idx, tuple(im.shape), lb))

plt.figure(figsize=(4, 4))
plt.imshow(im.squeeze(0), cmap="gray")
plt.title("idx=%d  label=%d  shape=%s" % (idx, lb, tuple(im.shape)))
plt.axis("off")
plt.tight_layout()
f_one = "one_image_%d.png" % SID
plt.savefig(f_one, dpi=110)
print("saved:", f_one)
plt.show()

# 🟢 EDIT: turn the image into its 784 pixel values (0-255 integers). Hint: im.squeeze(0) drops the channel, then convert to NumPy, ×255, and use round() to round
# YOUR CODE HERE
pixels = np.round(im.squeeze(0).numpy() * 255).astype(int)

np.set_printoptions(linewidth=260, threshold=10000)
print(pixels)
print("\n%d numbers in total (%d x %d)" % (pixels.size, pixels.shape[0], pixels.shape[1]))
```

## Question 3 · Take one batch and change `batch_size`

Take one batch with the DataLoader and print the shapes, then choose your own `batch_size` and count how many batches the dataset is split into.


```python
dl = DataLoader(ds, batch_size=64, shuffle=True)
xb, yb = next(iter(dl))
print("shape of one batch of images =", xb.shape, " labels =", yb.shape)
print("pixel range: %.3f to %.3f" % (float(xb.min()), float(xb.max())))

# 🟡 READ: the four batch sizes you may choose from
BS_TABLE = [8, 16, 32, 64]

# 🟢 EDIT: pick one, build the DataLoader, then count the batchesset 
# BS = BS_TABLE[SID % 4], build dl2 by copying the dl line above and replacing 64 with BS, then count n_batches = len(dl2).
# YOUR CODE HERE
BS = BS_TABLE[SID % 4]
dl2 = DataLoader(ds, batch_size=BS, shuffle=True)
n_batches = len(dl2)

print("your batch_size =", BS)
print("len(dl2) =", n_batches)
print("check: ceil(60000 / %d) = %d" % (BS, math.ceil(len(ds) / BS)))
```
