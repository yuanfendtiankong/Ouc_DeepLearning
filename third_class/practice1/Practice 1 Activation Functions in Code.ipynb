## Question 1 · Implement the three activations and their derivatives

Write the six functions, then draw the three curves together with their derivatives and save it.
The figure name must contain your `SID`.

```python
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
plt.rcParams["font.sans-serif"] = ["DejaVu Sans"]
plt.rcParams["axes.unicode_minus"] = False

# 🟢 EDIT: set SID to the last two digits of your student ID
SID = 23


def sigmoid(x):
    """1 / (1 + exp(-x))"""
    # YOUR CODE HERE
    return 1 / (1 + np.exp(-x))


def d_sigmoid(x):
    """s(x) * (1 - s(x))"""
    # YOUR CODE HERE
    return sigmoid(x) *(1- sigmoid(x))


def tanh(x):
    """(exp(x) - exp(-x)) / (exp(x) + exp(-x))"""
    # YOUR CODE HERE
    return (np.exp(x) - np.exp(-x)) / (np.exp(x) + np.exp(-x))


def d_tanh(x):
    # YOUR CODE HERE
    return 1- np.tanh(x) ** 2


def relu(x):
    """max(0, x)"""
    # YOUR CODE HERE
    return np.maximum(0,x)


def d_relu(x):
    """1 where x > 0, otherwise 0"""
    # YOUR CODE HERE
    return np.where(np.asanyarray(x)>0,1.0,0.0)


# ⚪ SKIP: check yourself
print("sigmoid(0) =", sigmoid(0.0), " tanh(0) =", tanh(0.0), " relu(0) =", relu(0.0))



# 🟡 READ: draw one panel per activation (function + derivative), then save the figure
# the grey bands are where the derivative is (almost) zero:
#          for sigmoid / tanh these are the saturated regions; for ReLU it is the whole negative half.
x = np.linspace(-6, 6, 400)
fig, axes = plt.subplots(1, 3, figsize=(11, 3.2))
acts = [("sigmoid", sigmoid, d_sigmoid, "#1F4E79"),
        ("tanh", tanh, d_tanh, "#2E8B57"),
        ("relu", relu, d_relu, "#C0392B")]
for ax, (name, f, d, col) in zip(axes, acts):
    if name == "relu":
        ax.axvspan(-6, 0, color="0.90")                 # dead zone: derivative is 0
    else:
        ax.axvspan(-6, -4, color="0.90")                # saturated region
        ax.axvspan(4, 6, color="0.90")
    ax.axhline(0, color="0.6", lw=0.8)
    ax.plot(x, f(x), color=col, lw=2.2, label="%s(x)" % name)
    ax.plot(x, d(x), color=col, lw=1.6, ls="--", label="%s'(x)" % name)
    ax.plot([0.0], [f(0.0)], "o", color=col, ms=6)
    ax.set_title("%s   f(0)=%.2f   max f'=%.2f" % (name, f(0.0), float(np.max(d(x)))), fontsize=10)
    ax.set_ylim(-1.2, 1.6); ax.set_xlim(-6, 6)
    ax.legend(fontsize=8, loc="lower right"); ax.grid(alpha=0.25)
plt.tight_layout()
plt.savefig("activation_%d.png" % SID, dpi=110)
plt.show()
```
