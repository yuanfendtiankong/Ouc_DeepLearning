Complete the following code to create the 'Gradient Descent' GIF from the class PPT and save the image.

Specifically, you need to record the gradient descent path by defining `xs`, `ys`, and `zs` as NumPy arrays. Each array should contain `num_iter` elements, starting from the initial point (`start_x`, `start_y`) and updating according to the gradient descent rule with learning rate `lr`.



```python

import numpy as np
import matplotlib

matplotlib.use('Agg')

import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter

# 🟡 READ	read it, no need to change
#===================== 1. Define the loss function and gradient =====================

def loss_func(x, y):
    #Take a bowl-shaped convex surface as an example to explain
    return x**2 + y**2


def grad(x, y):
    dx = 2 * x
    dy = 2 * y
    return dx, dy


# 🟡 READ	read it, no need to change
#===================== 2. gradient descent parameters =====================

lr = 0.1

# Starting point of the ball
start_x, start_y = -3, 3

# Number of iterations
num_iter = 30


# 🟢 EDIT	change this line
#===================== 3. Record the gradient descent path =====================

# YOUR CODE HERE
xs = np.zeros(num_iter)
ys = np.zeros(num_iter)
zs = np.zeros(num_iter)

x, y = start_x, start_y
for i in range(num_iter):
    xs[i] = x
    ys[i] = y
    zs[i] = loss_func(x, y)

    dx,dy = grad(x,y)
    x = x - lr * dx
    y = y - lr * dy



# 🟡 READ	read it, no need to change
#===================== 4. Initialize canvas =====================

fig = plt.figure(figsize=(10, 8))

ax = fig.add_subplot(111, projection='3d')


# 🟡 READ	read it, no need to change
#===================== 5. Generate loss function surface =====================

X_grid, Y_grid = np.meshgrid(
    np.linspace(-4, 4, 80),
    np.linspace(-4, 4, 80)
)

Z_grid = loss_func(X_grid, Y_grid)


# Plot the loss surface
surf = ax.plot_surface(
    X_grid,
    Y_grid,
    Z_grid,
    cmap='viridis',
    alpha=0.6,
    edgecolor='#444444',
    linewidth=0.2
)


# 🟡 READ	read it, no need to change
#===================== 6. Initialize the ball =====================

point, = ax.plot(
    [],
    [],
    [],
    'ro',
    markersize=10
)


# 🟡 READ	read it, no need to change
#===================== 7. Initialize trajectory =====================

trail, = ax.plot(
    [],
    [],
    [],
    'r-',
    linewidth=2
)


# 🟡 READ	read it, no need to change
#===================== 8. Set up the axes =====================

ax.set_xlim(-4, 4)
ax.set_ylim(-4, 4)
ax.set_zlim(0, 20)

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('Loss')

ax.set_title(
    "Gradient Descent Animation: Ball Rolling on Loss Surface"
)


# 🟡 READ	read it, no need to change
#===================== 9. Animation Update Function =====================

def update(frame):

    # The current position of the ball
    point.set_data(
        [xs[frame]],
        [ys[frame]]
    )

    point.set_3d_properties(
        [zs[frame]]
    )

    # Update Track
    trail.set_data(
        xs[:frame + 1],
        ys[:frame + 1]
    )

    trail.set_3d_properties(
        zs[:frame + 1]
    )

    return point, trail


# 🟡 READ	read it, no need to change
#===================== 10. Create animation =====================

ani = FuncAnimation(
    fig,
    update,
    frames=num_iter,
    interval=300,
    blit=False
)


# 🟡 READ	read it, no need to change
#===================== 11. Save GIF =====================

output_file = "gradient_descent.gif"

import os

# Autograding only checks xs / ys / zs, so the GIF is skipped there to keep grading fast.
# When you run this cell yourself, the GIF is generated as usual.
if os.environ.get("NBGRADER_EXECUTION") == "autograde":
    print("(Autograding: GIF generation skipped)")
else:
    print("Generating a gradient descent GIF, please wait...")

    ani.save(
        output_file,
        writer=PillowWriter(fps=3)
    )

    print("====================================")
    print("Animation generated successfully!")
    print("Documents:", output_file)
    print("====================================")


# Close the canvas
plt.close(fig)
```

## Question 2 · Numeric Gradient

Given the function:

$$
f(w) = w_0^2 + 3w_1^2 + 2w_2^2
$$

The numeric gradient is computed by perturbing each parameter slightly, recalculating the function value, and using the central difference:

$$
\text{grad}[i] = \frac{f(w + h \cdot e_i) - f(w - h \cdot e_i)}{2h}
$$

where $e_i$ is the unit vector in the $i$-th dimension.

Please complete the function `numeric_gradient(w, h=1e-5)` in the answer cell below.

Requirements:
1. Perturb each parameter of `w` individually.
2. Compute the function values before and after perturbation.
3. Use the central difference formula above.
4. Return a gradient array with the same shape as `w`.


```python

# 🟡 READ	read it, no need to change

import numpy as np

def loss_function(w):
    """
    f(w) = w0^2 + 3*w1^2 + 2*w2^2
    """
    return w[0]**2 + 3*w[1]**2 + 2*w[2]**2


def numeric_gradient(w, h=1e-5):
    """
    Compute the numeric gradient of loss_function at w using central difference.
    
    Parameters:
    w: numpy array of shape (3,)
    h: step size for perturbation
    
    Returns:
    grad: numpy array of the same shape as w
    """
    # 🟢 EDIT	change this line 
    #===========Compute the gradient of the loss with respect to w using numerical methods===========
    
    # YOUR CODE HERE
    w = np.asarray(w,dtype=float)
    grad = np.zeros_like(w)

    for i in range(len(w)):
        w1 = w.copy()
        w2 = w.copy()

        w1[i]+=h
        w2[i]-=h

        f1 = loss_function(w1)
        f2 = loss_function(w2)
        grad[i] = (f1 - f2) / (2 * h)

    return grad
```
