## Question 1 · Batch Bounding Box Regression

Based on the "R-CNN: Bounding-Box Regression" slides, the model predicts a transform $(t_x, t_y, t_w, t_h)$ to correct the region proposal $(p_x, p_y, p_w, p_h)$.
The final bounding box $(b_x, b_y, b_w, b_h)$ is calculated as:

$$
b_x = p_x + p_w t_x
$$
$$
b_y = p_y + p_h t_y
$$
$$
b_w = p_w \exp(t_w)
$$
$$
b_h = p_h \exp(t_h)
$$

In practice, we process a batch of proposals simultaneously. Implement the function `apply_bbox_transform(proposals, deltas)` using `numpy` to perform this calculation for an entire batch.

- `proposals`: A numpy array of shape `(N, 4)`, where each row is `[px, py, pw, ph]`.
- `deltas`: A numpy array of shape `(N, 4)`, where each row is `[tx, ty, tw, th]`.
- Return: A numpy array of shape `(N, 4)` containing `[bx, by, bw, bh]` for each proposal.

**Hints**:
1. Extract the columns from the arrays: 
   `px, py, pw, ph = proposals[:, 0], proposals[:, 1], proposals[:, 2], proposals[:, 3]`
   `tx, ty, tw, th = deltas[:, 0], deltas[:, 1], deltas[:, 2], deltas[:, 3]`
2. Apply the formulas element-wise (NumPy will handle the batch dimension automatically). Use `np.exp()` for the scaling factors.
3. Stack the resulting arrays back into an `(N, 4)` matrix using `np.stack([bx, by, bw, bh], axis=1)`.


```python

# Grid Type: Autograded answer
import numpy as np

def apply_bbox_transform(proposals, deltas):
    # 🟢 EDIT: Extract the px, py, pw, ph and tx, ty, tw, th arrays
    # YOUR CODE HERE
    px, py, pw, ph = proposals[:, 0], proposals[:, 1], proposals[:, 2], proposals[:, 3]
    tx, ty, tw, th = deltas[:, 0], deltas[:, 1], deltas[:, 2], deltas[:, 3]
    
    # 🟢 EDIT: Calculate the transformed coordinates using the formulas
    # YOUR CODE HERE
    bx = px + pw * tx
    by = py + ph * ty
    bw = pw * np.exp(tw)
    bh = ph * np.exp(th)
    
    # 🟢 EDIT: Stack and return the result
    # YOUR CODE HERE
    return np.stack([bx, by, bw, bh], axis=1)
    
```


## Question 2 · Intersection over Union (IoU)

Based on the "NMS: Intersection over Union" slides, IoU is used to measure the overlap between two bounding boxes. It is defined as:

$$
\text{IoU} = \frac{\text{Area of Overlap}}{\text{Area of Union}}
$$

Implement the function `calculate_iou(boxA, boxB)`.
- `boxA` and `boxB` are lists or tuples in the format `[x1, y1, x2, y2]`, where `(x1, y1)` is the top-left corner and `(x2, y2)` is the bottom-right corner.
- Return the IoU value as a float. If there is no overlap, return `0.0`.

**Hints**:
1. Find the coordinates of the intersection rectangle:
   `x1_inter = max(boxA[0], boxB[0])`, `y1_inter = max(boxA[1], boxB[1])`
   `x2_inter = min(boxA[2], boxB[2])`, `y2_inter = min(boxA[3], boxB[3])`
2. Calculate the intersection area. If `x2_inter < x1_inter` or `y2_inter < y1_inter`, the area is 0.
3. Calculate the areas of both boxes: `areaA = (boxA[2] - boxA[0]) * (boxA[3] - boxA[1])`.
4. Union area = `areaA + areaB - inter_area`. Return `inter_area / union_area`.


```python
def calculate_iou(boxA, boxB):
    # 🟢 EDIT: Calculate intersection coordinates
    # YOUR CODE HERE
    x1_inter = max(boxA[0], boxB[0])
    y1_inter = max(boxA[1], boxB[1])
    x2_inter = min(boxA[2], boxB[2])
    y2_inter = min(boxA[3], boxB[3])
    
    # 🟢 EDIT: Calculate intersection area (handle no overlap case)
    # YOUR CODE HERE
    if x2_inter < x1_inter or y2_inter < y1_inter:
        inter_area = 0.0
    else:
        inter_area = (x2_inter - x1_inter) * (y2_inter - y1_inter)
    
    # 🟢 EDIT: Calculate union area and return IoU
    # YOUR CODE HERE
    areaA = (boxA[2] - boxA[0]) * (boxA[3] - boxA[1])
    areaB = (boxB[2] - boxB[0]) * (boxB[3] - boxB[1])
    union_area = areaA + areaB - inter_area
    if union_area == 0:
        return 0.0
    return float(inter_area / union_area)
    
```


## Question 3 · Precision and Recall

Based on the "Metric: mAP" slides, precision and Recall are defined as:

$$
\text{Precision} = \frac{TP}{TP + FP}
$$

$$
\text{Recall} = \frac{TP}{TP + FN}
$$

where TP is True Positive, FP is False Positive, and FN is False Negative.

Implement the function `calculate_metrics(tp, fp, fn)`.
Return a tuple `(precision, recall)`.
If the denominator is zero, return `0.0` for that metric.

**Hints**:
1. Calculate `precision`: check if `tp + fp == 0`.
2. Calculate `recall`: check if `tp + fn == 0`.


```python
def calculate_metrics(tp, fp, fn):
    # 🟢 EDIT: Calculate Precision
    # YOUR CODE HERE
    if tp + fp != 0:
        precision = tp / (tp + fp)
    else :
        precision = 0.0
    
    # 🟢 EDIT: Calculate Recall
    # YOUR CODE HERE
    if tp + fn != 0:
        recall = tp / (tp + fn)
    else:
        recall = 0.0
    
    return precision, recall
    
```
