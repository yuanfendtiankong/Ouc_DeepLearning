## Question 1 · Dot-Product Attention Mechanism

Attention allows the Decoder to focus on relevant parts of the Encoder outputs. 
A common variant is **Dot-Product Attention**.

Implement the function `dot_product_attention(Q, K, V)` using NumPy.
- `Q`: Query vector, shape `(d,)`.
- `K`: Key matrix, shape `(T, d)`.
- `V`: Value matrix, shape `(T, d_v)`.
- Return the context vector of shape `(d_v,)`.

**Formulas**:
1. Score: $\text{scores} = Q \cdot K^T$
2. Attention Weights: $\alpha = \text{softmax}(\text{scores})$
3. Context Vector: $\text{context} = \alpha \cdot V$

**Hints**:
1. Calculate `scores` using `np.dot(Q, K.T)`.
2. Compute `alpha` using a numerically stable softmax: `np.exp(scores - np.max(scores)) / np.sum(np.exp(scores - np.max(scores)))`.
3. Calculate `context` using `np.dot(alpha, V)`.


```python

import numpy as np

def dot_product_attention(Q, K, V):
    # 🟢 EDIT: Calculate the attention scores
    # YOUR CODE HERE
    scores = np.dot(Q, K.T)
    
    # 🟢 EDIT: Apply softmax to get attention weights
    # YOUR CODE HERE
    alpha = np.exp(scores - np.max(scores)) / np.sum(np.exp(scores - np.max(scores)))
    
    # 🟢 EDIT: Calculate the context vector
    # YOUR CODE HERE
    context = np.dot(alpha, V)
    
    return context
    
```


## Question 2 · RPN Anchor Label Assignment

Training the Region Proposal Network (RPN) requires assigning positive and negative labels to anchors based on their Intersection over Union (IoU) with Ground Truth boxes.

Implement the function `assign_anchor_labels(iou_array)` using NumPy.
- `iou_array`: A 1D NumPy array of shape `(N,)` containing the maximum IoU of each anchor with any Ground Truth box.
- Return an array `labels` of shape `(N,)` where:
  - Label `1`: Positive anchor (IoU >= 0.7)
  - Label `0`: Negative anchor (IoU < 0.3)
  - Label `-1`: Neutral/Ignore anchor (0.3 <= IoU < 0.7)

**Hints**:
1. Initialize a `labels` array with `-1` of the same length as `iou_array`.
2. Use boolean indexing to set `labels[iou_array < 0.3] = 0`.
3. Use boolean indexing to set `labels[iou_array >= 0.7] = 1`.


```python

import numpy as np

def assign_anchor_labels(iou_array):
    # 🟢 EDIT: Initialize an array with -1 (neutral)
    # YOUR CODE HERE
    labels = np.full(iou_array.shape, -1, dtype=int)
    
    # 🟢 EDIT: Assign negative labels (IoU < 0.3) and positive labels (IoU >= 0.7)
    # YOUR CODE HERE
    labels[iou_array < 0.3] = 0
    labels[iou_array >= 0.7] = 1
    
    return labels
    
```


## Question 3 · Compare R-CNN and YOLO Classification Heads

Based on the provided slides, **R-CNN** is a Two-stage detector. It first extracts $N$ Region Proposals and passes each proposal through a Fully Connected (FC) layer for classification.
**YOLO** is a One-stage detector. It divides the image into an $S \times S$ grid and uses a single Convolutional layer to predict the class scores for all grid cells simultaneously.

Given an input feature dimension `D`, number of classes `C`, number of proposals `N`, and grid size `S`, implement the function `compare_heads(D, C, N, S)` to calculate the number of parameters for the classification head of each method.

**Formulas**:
- **R-CNN Head**: Parameters per proposal = $D \times C + C$. Total parameters = $N \times (D \times C + C)$.
- **YOLO Head**: A 1x1 Conv layer outputs $S \times S \times C$ features. Parameters = $D \times (S \times S \times C) + (S \times S \times C)$.

Return a dictionary `{"R-CNN": rcnn_params, "YOLO": yolo_params}`.

**Hints**:
1. For R-CNN, calculate the FC layer parameters and multiply by `N`.
2. For YOLO, calculate the total output elements first (`S * S * C`).
3. Calculate the convolution parameters using `D` and the total output elements.



```python

def compare_heads(D, C, N, S):
    # 🟢 EDIT: Calculate R-CNN classification head parameters
    # YOUR CODE HERE
    rcnn_params = N * (D * C + C)
    
    # 🟢 EDIT: Calculate YOLO classification head parameters
    # YOUR CODE HERE
    yolo_outputs = S * S * C
    yolo_params = D * yolo_outputs + yolo_outputs
    
    return {"R-CNN": rcnn_params, "YOLO": yolo_params}
```
