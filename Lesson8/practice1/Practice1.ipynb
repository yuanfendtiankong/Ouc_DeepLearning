## Question 1 · Sliding Window Number of Boxes

If we use a sliding window approach to detect objects, the number of possible boxes in an image of size $H \times W$ is given by the formula:

$$
\text{Total Boxes} = \frac{H(H + 1)}{2} \times \frac{W(W + 1)}{2}
$$

Implement the function `count_sliding_windows(H, W)` using Python to calculate this total number of boxes.

**Hints**:
To solve this, you should first compute the number of combinations for the height using the formula `H * (H + 1) // 2`, and then compute the combinations for the width using `W * (W + 1) // 2`. Finally, multiply these two intermediate results together to obtain and return the final value.


```python
def count_sliding_windows(H, W):
    # 🟢 EDIT: Calculate the total number of boxes using the formula from the hints.
    # YOUR CODE HERE
    return (H + 1) * H // 2 * (W + 1) * W // 2
    
```


## Question 2 · Multitask Loss for Object Detection

Detecting a single object can be treated as a regression problem for the bounding box coordinates \((x, y, w, h)\) and a classification problem for the label. 
The final loss is a **Weighted Sum** of the classification loss (Softmax Loss) and the bounding box regression loss (L2 Loss).

Implement the function `compute_multitask_loss(pred_box, target_box, cls_loss, reg_weight=1.0)` using NumPy.
- `pred_box`: Predicted box coordinates `(x, y, w, h)`, shape `(4,)`.
- `target_box`: Ground truth box coordinates `(x, y, w, h)`, shape `(4,)`.
- `cls_loss`: A float representing the classification loss.
- `reg_weight`: A float representing the weight for the regression loss.

The total loss formula is:
$$
 L = L_C + \lambda \cdot L_R 
$$
where $L_C$ is `cls_loss`, $L_R$ is the L2 regression loss, and $\lambda$ is `reg_weight`.

**Hints**:
1. The L2 regression loss $L_R$ is the **sum of squared differences** between the predicted and target coordinates. You can use `np.sum((pred_box - target_box) ** 2)` to calculate it.
2. The total loss $L$ is the classification loss plus the weighted regression loss. 


```python
import numpy as np

def compute_multitask_loss(pred_box, target_box, cls_loss, reg_weight=1.0):
    # 🟢 EDIT: Calculate the L2 regression loss for the box coordinates
    # YOUR CODE HERE
    LR = np.sum((pred_box - target_box)**2)
    
    # 🟢 EDIT: Calculate the weighted sum of classification and regression losses
    # YOUR CODE HERE
    total_loss = cls_loss + reg_weight * LR
    
    return total_loss
    
```
