### I. Background

We have learned the basic structure of neural networks and the forward propagation process. In this question, we implement a **two-layer fully connected neural network**:

$$
z_1 = x W_1 + b_1
$$
$$
a_1 = \mathrm{ReLU}(z_1) = \max(0, z_1)
$$
$$
y = a_1 W_2 + b_2
$$

The first layer uses ReLU activation, and the second layer is purely linear (no activation).

### II. Task Requirements

Please complete the `forward(x, W1, b1, W2, b2)` function in the next cell.

Requirements:

1. Compute the first linear transformation `z1 = x @ W1 + b1`.
2. Apply ReLU activation: `a1 = np.maximum(0, z1)`.
3. Compute the second linear transformation `y = a1 @ W2 + b2`.
4. Return the final output `y`.
5. Use NumPy and matrix multiplication; **support both single sample and batch input**.
6. There is no need to implement backpropagation.



```python

import numpy as np

# 🟢 EDIT: Implementing the forward pass of a two-layer fully connected neural network
def forward(x, W1, b1, W2, b2):
    """
    Forward propagation of a two-layer fully connected neural network.
    Layer 1: linear + ReLU
    Layer 2: linear (no activation)

    Parameters:
        x:  input vector or batch, shape (n,) or (batch, n)
        W1: weight matrix of the hidden layer, shape (n, h)
        b1: bias vector of the hidden layer, shape (h,)
        W2: weight matrix of the output layer, shape (h, m)
        b2: bias vector of the output layer, shape (m,)

    Returns:
        y: output after the second linear transformation, shape (m,) or (batch, m)
    """
    # YOUR CODE HERE
    hidden = np.maximum(x @ W1 + b1, 0)
    return hidden @ W2 + b2
    
```
