## Question 1 · Scale-Invariant Signal-to-Distortion Ratio (SI-SDR)

SI-SDR measures the quality of separated speech by projecting the ground truth signal onto the estimated signal.
The formulas are:

$$
X_T = \frac{X^* \cdot \hat{X}}{\|\hat{X}\|^2} \hat{X}
$$
$$
X_E = X^* - X_T
$$
$$
\text{SI-SDR} = 10 \log_{10} \frac{\|X_T\|^2}{\|X_E\|^2 + \epsilon}
$$

Implement `calculate_si_sdr(X_star, X_hat)` using NumPy. Add a small epsilon (`1e-8`) to the denominator to avoid division by zero.

**Task**: Fill in the missing blocks to complete the calculation.

**Tip!！**: When calculating np.log10, also add `1e-8` to the numerator to avoid getting negative infinity and RuntimeWarning

```python

# Grid Type: Autograded answer
import numpy as np

def calculate_si_sdr(X_star, X_hat):
    # 🟢 EDIT: Compute the scaling factor alpha
    # YOUR CODE HERE
    X_star = np.asarray(X_star, dtype=float)
    X_hat = np.asarray(X_hat, dtype=float)
    epsilon = 1e-8
    alpha = np.dot(X_star, X_hat) / (np.sum(X_hat ** 2) + epsilon)
    
    # 🟢 EDIT: Compute the target projection X_T and the error X_E
    # YOUR CODE HERE
    X_T = alpha * X_hat
    X_E = X_star - X_T
    
    # 🟢 EDIT: Compute the final SI-SDR value
    # YOUR CODE HERE
    si_sdr = 10 * np.log10(
        (np.sum(X_T ** 2) + epsilon) / (np.sum(X_E ** 2) + epsilon)
    )
    
    return si_sdr
    
```


## Question 2 · Speaker Verification with Cosine Similarity

The system extracts speaker embeddings and calculates their similarity to make a verification decision (Same or Different).

Implement `verify_speaker(emb1, emb2, threshold)` using the Cosine Similarity formula:

$$
\text{similarity} = \frac{A \cdot B}{\|A\| \|B\|}
$$

Return `True` if similarity $\ge$ threshold, otherwise `False`. If either embedding is a zero vector, return `False`.

**Task**: Fill in the missing blocks to compute the similarity and make the decision.


```python

# Grid Type: Autograded answer
import numpy as np

def verify_speaker(emb1, emb2, threshold):
    v1, v2 = np.array(emb1), np.array(emb2)
    norm1, norm2 = np.linalg.norm(v1), np.linalg.norm(v2)
    
    # 🟢 EDIT: Handle zero vectors
    # YOUR CODE HERE
    if norm1 == 0 or norm2 == 0:
        return False
    
    # 🟢 EDIT: Compute cosine similarity
    # YOUR CODE HERE
    similarity = np.dot(v1.astype(float), v2.astype(float)) / (norm1 * norm2)
    
    # 🟢 EDIT: Return the verification result
    # YOUR CODE HERE
    return bool(similarity >= threshold)
    
```


## Question 3 · Equal Error Rate (EER) Calculation

The Equal Error Rate (EER) is the point where the False Positive Rate (FPR) equals the False Negative Rate (FNR). 

Given arrays of thresholds, FPR, and FNR, implement `calculate_eer(fpr, fnr, thresholds)`:
1. Find the index where the absolute difference between `fpr` and `fnr` is minimized.
2. Return a tuple `(best_threshold, eer_value)` where `eer_value` is the average of `fpr` and `fnr` at that index.

**Task**: Fill in the missing blocks to complete the calculation.


```python

# Grid Type: Autograded answer
import numpy as np

def calculate_eer(fpr, fnr, thresholds):
    fpr, fnr = np.array(fpr), np.array(fnr)
    
    # 🟢 EDIT: Find the index where the absolute difference is minimized
    # YOUR CODE HERE
    best_index = np.argmin(np.abs(fpr - fnr))
    
    # 🟢 EDIT: Extract the best threshold and calculate EER
    # YOUR CODE HERE
    best_threshold = thresholds[best_index]
    eer_value = (fpr[best_index] + fnr[best_index]) / 2
    
    return best_threshold, eer_value
```
