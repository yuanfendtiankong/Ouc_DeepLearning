## Question 1 · Audio Framing

Raw audio signals must be divided into short overlapping frames before further processing (e.g., 25ms window with 10ms hop). 

Implement the function `framing(signal, frame_size, hop_size)` that splits a 1D signal into a 2D array of frames. 
- `signal`: 1D numpy array.
- `frame_size`: Number of samples per frame.
- `hop_size`: Number of samples between consecutive frames.
- Return: A 2D numpy array of shape `(num_frames, frame_size)`.

**Task**: Fill in the missing blocks to initialize the matrix and extract the frames.


```python

import numpy as np

def framing(signal, frame_size, hop_size):
    num_frames = 1 + (len(signal) - frame_size) // hop_size
    
    # 🟢 EDIT: Initialize the output matrix
    # YOUR CODE HERE
    frames = np.zeros((num_frames, frame_size), dtype=signal.dtype)
    
    for i in range(num_frames):
        start = i * hop_size
        end = start + frame_size
        
        # 🟢 EDIT: Extract the frame from the signal
        # YOUR CODE HERE
        frames[i] = signal[start:end]
        
    return frames
    
```


## Question 2 · Pre-emphasis Filter

Pre-emphasis is a common preprocessing step in speech processing to compensate for the high-frequency part of the speech signal that is suppressed during human sound production. The filter is defined as:

$$
y[n] = x[n] - \alpha x[n-1]
$$

where $x$ is the input signal and $\alpha$ is the pre-emphasis coefficient.

Implement `pre_emphasis(signal, alpha)` using NumPy vectorization.
- `signal`: 1D numpy array.
- `alpha`: Float, the coefficient.
- Return: The filtered 1D numpy array.

**Task**: Fill in the missing blocks to compute the first element and the rest of the vectorized operations.


```python

import numpy as np

def pre_emphasis(signal, alpha=0.97):
    y = np.zeros_like(signal)
    
    # 🟢 EDIT: The first element remains unchanged
    # YOUR CODE HERE
    y = y.astype(np.result_type(signal.dtype, alpha))
    if signal.size > 0:
        y[0] = signal[0]
    
    # 🟢 EDIT: Vectorized computation for the rest
    # YOUR CODE HERE
    y[1:] = signal[1:] - alpha * signal[:-1]
    
    return y
    
```


## Question 3 · Windowing

After framing, a window function (like the Hamming window) is multiplied with each frame to reduce spectral leakage. The Hamming window of length $N$ is defined as:

$$
w[n] = 0.54 - 0.46 \cos\left(\frac{2\pi n}{N-1}\right), \quad 0 \le n \le N-1
$$

Implement `apply_hamming_window(frame)`.
- `frame`: A 1D numpy array representing a single audio frame.
- Return: The frame multiplied by the Hamming window.

**Task**: Fill in the missing blocks to generate the window sequence and apply it.


```python

# Grid Type: Autograded answer
import numpy as np

def apply_hamming_window(frame):
    N = len(frame)
    
    # 🟢 EDIT: Generate the Hamming window sequence
    # YOUR CODE HERE
    n = np.arange(N)
    window = np.ones(N) if N == 1 else 0.54 - 0.46 * np.cos(2 * np.pi * n / (N - 1))
    
    # 🟢 EDIT: Multiply the frame by the window
    # YOUR CODE HERE
    result = frame * window
    
    return result
    
```


## Question 4 · Voice Activity Detection via Short-Time Energy

In audio preprocessing, removing silent segments (Voice Activity Detection) is a crucial step. A standard method is to compute the **Short-Time Energy (STE)** for each frame of the signal and apply a threshold.

The Short-Time Energy for a frame $x_m$ is defined as the mean of the squared values: 

$$
E_m = \frac{1}{N} \sum_{n=0}^{N-1} x_m[n]^2
$$

Implement `detect_speech_segments(signal, frame_size, hop_size, threshold)`:
1. Split the `signal` into frames using `frame_size` and `hop_size`.
2. Calculate the energy for each frame.
3. Find all frames where the energy is strictly greater than `threshold`.
4. Return a tuple `(start_idx, end_idx)` representing the index of the first and last active frames. If no frames exceed the threshold, return `(-1, -1)`.

**Task**: Fill in the missing blocks to extract the frame, calculate energy, find active indices, and return the result.



```python

# Grid Type: Autograded answer
import numpy as np

def detect_speech_segments(signal, frame_size, hop_size, threshold):
    num_frames = 1 + (len(signal) - frame_size) // hop_size
    energy = np.zeros(num_frames)
    
    for i in range(num_frames):
        # 🟢 EDIT: Extract the frame and compute the mean squared energy
        # YOUR CODE HERE
        start = i * hop_size
        frame = signal[start:start + frame_size].astype(float)
        energy[i] = np.mean(frame ** 2)
        
    # 🟢 EDIT: Find frame indices where energy is greater than the threshold
    # YOUR CODE HERE
    active_frames = np.flatnonzero(energy > threshold)
    
    # 🟢 EDIT: Return the first and last active frame indices, or (-1, -1)
    # YOUR CODE HERE
    if active_frames.size == 0:
        return (-1, -1)
    return (int(active_frames[0]), int(active_frames[-1]))
    
```
