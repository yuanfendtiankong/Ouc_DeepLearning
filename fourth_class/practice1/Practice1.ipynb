## Question 1 · MLP for Image Classification

When an image is fed into a Multi-Layer Perceptron (MLP), it must be flattened into a 1D column vector ("Stretch pixels into column"). This operation destroys the spatial structure and often leads to a massive number of parameters.

Given an image of shape `(C, H, W)`, implement the following two functions using Python:

1. `flatten_image_size(C, H, W)`: Return the size of the flattened 1D vector.
2. `calculate_mlp_params(input_size, hidden_size, num_classes)`: Calculate the total number of parameters (weights + biases) for a 1-hidden-layer MLP. The architecture is:
   - Input layer of size `input_size`
   - Hidden layer of size `hidden_size` (with bias)
   - Output layer of size `num_classes` (with bias)

Return the calculated values as integers.


```python

def flatten_image_size(C, H, W):
    """Calculate the size of the flattened image vector."""
    # 🟢 EDIT: change this line (Think about the total number of elements in the 3D tensor)
    return C * H * W

def calculate_mlp_params(input_size, hidden_size, num_classes):
    """Calculate total parameters for a 1-hidden-layer MLP."""
    # 🟢 EDIT: change this line (Sum the weights and biases for both layers)
    return input_size * hidden_size + hidden_size + hidden_size * num_classes + num_classes
    
```


## Question 2 · Valid 2D Convolution with Shared Weights

The slides illustrate convolution as shared connections / spatial filtering. Implement `conv2d_valid(image, kernel)` for a single-channel image and a single 2D kernel. Return the valid convolution output. You may use numpy.

**Task**: Fill in the missing parts of the function skeleton to complete the valid 2D convolution. You need to:
1. Compute the output height and width.
2. Initialize the output matrix with zeros.
3. Use nested loops to slide the kernel over the image.
4. Compute the convolution value for each output position.

The function framework (input conversion, shape retrieval, and return) is already provided.


```python

import numpy as np

def conv2d_valid(image, kernel):
    """Return the valid 2D convolution of image with kernel."""
    image = np.asarray(image)
    kernel = np.asarray(kernel)
    H, W = image.shape
    KH, KW = kernel.shape
    
    # 🟢 EDIT: change this line (Compute the output size, initialize a zero matrix, then slide the kernel to compute the sum of element-wise products)
    output_height = H - KH + 1
    output_width = W - KW + 1
    output = np.zeros((output_height, output_width), dtype=np.result_type(image, kernel))

    for i in range(output_height):
        for j in range(output_width):
            window = image[i:i + KH, j:j + KW]
            output[i, j] = np.sum(window * kernel)

    return output
    
```


## Question 3 · MLP for Image Flattening and Parameter Calculation

Based on the provided image "MLPs for Image", when an image of size (32, 32, 3) is flattened into a column vector, it loses its spatial structure. Additionally, the parameter count explodes. 

Please implement the following functions:
1. `calculate_mlp_params(input_dim, hidden_dim, output_dim)`: Calculate the total number of parameters (weights + biases) in a 1-hidden-layer MLP.
2. `flatten_spatial_distance(H, W, p1, p2)`: Given a 2D image of size (H, W) flattened into a 1D array (row-major), return the absolute distance in the 1D array between pixel `p1=(r1, c1)` and `p2=(r2, c2)`.


```python

def calculate_mlp_params(input_dim, hidden_dim, output_dim):
    """Calculate the total number of parameters in a 1-hidden-layer MLP."""
    # 🟢 EDIT: change this line (Think about weight matrix sizes and bias counts for each layer)
    return input_dim * hidden_dim + hidden_dim + hidden_dim * output_dim + output_dim

def flatten_spatial_distance(H, W, p1, p2):
    """Return the absolute distance in the flattened array between p1 and p2.
    p1 = (r1, c1), p2 = (r2, c2)
    """
    # 🟢 EDIT: change this line (How to map 2D coordinates (r, c) to a 1D index in row-major order?)
    r1, c1 = p1
    r2, c2 = p2
    index1 = r1 * W + c1
    index2 = r2 * W + c2
    return abs(index1 - index2)
    
```
