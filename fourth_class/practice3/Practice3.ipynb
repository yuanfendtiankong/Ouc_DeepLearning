## Question 1 · Batch Normalization and Layer Normalization

Based on the Normalization slides, Batch Normalization and Layer Normalization differ in the dimensions over which they compute statistics.

Implement the following functions using NumPy:
1. `batch_norm_forward(x, gamma, beta, eps)`: 
   - Input `x` has shape `(N, C, H, W)`.
   - Calculate the per-channel mean and variance across `(N, H, W)`.
   - Return the normalized and scaled output with the same shape as `x`.
2. `layer_norm_forward(x, gamma, beta, eps)`:
   - Input `x` has shape `(N, C, H, W)`.
   - Calculate the mean and variance for each sample across `(C, H, W)`.
   - Return the normalized and scaled output with the same shape as `x`.

**Task:**
The shape extraction and reshaping of `gamma` and `beta` are already provided. Please fill in the missing code to:
1. Compute the mean and variance along the correct axes (use `keepdims=True`).
2. Normalize the input: $ \hat{x} = \frac{x - \mu}{\sqrt{\sigma^2 + \epsilon}} $
3. Scale and shift the normalized output: $ y = \gamma \hat{x} + \beta $


```python

import numpy as np

def batch_norm_forward(x, gamma, beta, eps=1e-5):
    """
    Forward pass for Batch Normalization.
    x: (N, C, H, W)
    gamma: (C,) or (1, C, 1, 1)
    beta: (C,) or (1, C, 1, 1)
    """
    N, C, H, W = x.shape
    gamma = gamma.reshape(1, C, 1, 1)
    beta = beta.reshape(1, C, 1, 1)
    
    # 🟢 EDIT: change this line (Compute per-channel statistics over (N, H, W), then normalize and scale)
    mean = np.mean(x, axis=(0, 2, 3), keepdims=True)
    variance = np.var(x, axis=(0, 2, 3), keepdims=True)
    x_hat = (x - mean) / np.sqrt(variance + eps)
    out = gamma * x_hat + beta
    return out

def layer_norm_forward(x, gamma, beta, eps=1e-5):
    """
    Forward pass for Layer Normalization.
    x: (N, C, H, W)
    gamma: (C, H, W) or (1, C, H, W)
    beta: (C, H, W) or (1, C, H, W)
    """
    N, C, H, W = x.shape
    gamma = gamma.reshape(1, C, H, W)
    beta = beta.reshape(1, C, H, W)
    
    # 🟢 EDIT: change this line (Compute per-sample statistics over (C, H, W), then normalize and scale)
    mean = np.mean(x, axis=(1, 2, 3), keepdims=True)
    variance = np.var(x, axis=(1, 2, 3), keepdims=True)
    x_hat = (x - mean) / np.sqrt(variance + eps)
    out = gamma * x_hat + beta
    return out
    
```


## Question 3 · Well Known CNN Architectures

Based on the "Well Known CNNs Architectures" slides, different architectures have distinct structural designs:

1. **AlexNet**: The first Conv layer uses a `11x11` kernel with `stride=4` and no padding. 
2. **Inception**: It concatenates the output channels from multiple parallel branches (e.g., 1x1, 3x3, 5x5 convolutions, and max pooling).
3. **ResNet**: It introduces a residual connection (skip connection) where the final output is `F(x) + x` to mitigate gradient vanishing.

**Task:**
Implement the following three functions:
1. `alexnet_conv1_size(H)`: Calculate the output spatial dimension (height) after AlexNet's first conv layer, given the input height `H`.
2. `inception_out_channels(c1, c3, c5, c_pool)`: Calculate the total number of output channels after concatenating the four branches.
3. `resnet_block(F_x, x)`: Compute the output of a residual block.

Use NumPy for the ResNet task.


```python

import numpy as np

def alexnet_conv1_size(H):
    """Calculate Conv1 output size for AlexNet (K=11, S=4, P=0)."""
    # 🟢 EDIT: change this line (Use the standard convolution output formula with K=11, S=4, P=0)
    return int((H - 11) // 4 + 1)

def inception_out_channels(c1, c3, c5, c_pool):
    """Calculate total output channels for an Inception module."""
    # 🟢 EDIT: change this line (Sum the channel numbers from all four parallel branches)
    return int(c1 + c3 + c5 + c_pool)

def resnet_block(F_x, x):
    """Compute the residual connection output."""
    # 🟢 EDIT: change this line (Add the identity input x directly to the output of the weight layers)
    return np.add(F_x, x)
```
